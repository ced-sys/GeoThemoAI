```python
import ee
import folium
import geemap
```

```python
ee.Authenticate()
```

```python
ee.Initialize(project='subterra-466121')
```

```python
dataset=ee.ImageCollection('GOOGLE/SATELLITE_EMBEDDING/V1/ANNUAL')
```

```python
point=ee.Geometry.Point(36.2694, -0.8867)
```

```python
image1=(dataset
        .filterDate('2023-01-01', '2024-01-01')
        .filterBounds(point)
        .first())
```

```python
image2=(dataset
        .filterDate('2024-01-01', '2025-01-01')
        .filterBounds(point)
        .first())
```

```python
vis_params={
    'min':-0.3,
    'max':0.3,
    'bands':['A01', 'A16', 'A09']
}
```

```python
Map=geemap.Map()
```

```python
Map.add_ee_layer(image1, vis_params, '2023 embeddings')
Map.add_ee_layer(image2, vis_params, '2024 embeddings')
```

```python
dot_prod=(image1
          .multiply(image2)
          .reduce(ee.Reducer.sum()))
```

```python
Map.add_ee_layer(
    dot_prod,
    {'min':0, 'max':1, 'palette':['white', 'black']},
    'Similarity between years (brighter=less similar)'
)
```

```python
Map.centerObject(point, 12)
```

```python
Map.add_basemap('SATELLITE')
```
